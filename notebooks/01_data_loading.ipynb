# 01 - Data Loading and Initial Exploration

## Introduction
This notebook handles the initial loading of the Household Power Consumption dataset and provides an overview of its structure.

### Dataset Description
The **Individual Household Electric Power Consumption Dataset** contains measurements of electric power consumption in one household with a one-minute sampling rate over a period of almost 4 years (Dec 2006 - Nov 2010).

**Features:**
- **Date**: Date in format dd/mm/yyyy
- **Time**: Time in format hh:mm:ss
- **Global_active_power**: Household global minute-averaged active power (kilowatt)
- **Global_reactive_power**: Household global minute-averaged reactive power (kilowatt)
- **Voltage**: Minute-averaged voltage (volt)
- **Global_intensity**: Household global minute-averaged current intensity (ampere)
- **Sub_metering_1**: Energy sub-metering No. 1 (watt-hour) - Kitchen
- **Sub_metering_2**: Energy sub-metering No. 2 (watt-hour) - Laundry room
- **Sub_metering_3**: Energy sub-metering No. 3 (watt-hour) - Water heater and AC

**Note:** Missing values are represented by '?'

```python
# Import required libraries
import pandas as pd
import numpy as np
import os
import warnings
warnings.filterwarnings('ignore')

print("Libraries imported successfully!")
print(f"Pandas version: {pd.__version__}")
print(f"NumPy version: {np.__version__}")
```

```python
# Define file paths
DATA_PATH = os.path.join('..', 'data', 'raw', 'household_power_consumption.txt')
PROCESSED_PATH = os.path.join('..', 'data', 'processed')

# Verify file exists
if os.path.exists(DATA_PATH):
    file_size = os.path.getsize(DATA_PATH) / (1024 * 1024)  # Size in MB
    print(f"Data file found: {DATA_PATH}")
    print(f"File size: {file_size:.2f} MB")
else:
    print(f"Data file not found at: {DATA_PATH}")
```

```python
# Load the dataset

df = pd.read_csv(
    DATA_PATH,
    sep=';',
    low_memory=False,
    na_values=['?', 'NA', 'null']
)

# Combine Date and Time columns into a single datetime column
df['datetime'] = pd.to_datetime(df['Date'] + ' ' + df['Time'], format='%d/%m/%Y %H:%M:%S')
df = df.drop(['Date', 'Time'], axis=1)

# Reorder columns to put datetime first
cols = ['datetime'] + [col for col in df.columns if col != 'datetime']
df = df[cols]

print(f"Dataset loaded successfully!")
print(f"Shape: {df.shape[0]:,} rows × {df.shape[1]} columns")
```

```python
# Display basic information
print("DATASET OVERVIEW\n")

print("1. First few rows:")
display(df.head(10))

print("\n2. Last few rows:")
display(df.tail(10))
```

```python
# Data types and memory usage
print("3. Data Types")
print(df.info())
```

```python
# Statistical summary
print("5. Statistical Summary:")
display(df.describe())
```

```python
# Check for missing values
print("6. Missing Values Analysis:")
missing_data = pd.DataFrame({
    'Column': df.columns,
    'Missing_Count': df.isnull().sum(),
    'Missing_Percentage': (df.isnull().sum() / len(df)) * 100
})
missing_data = missing_data[missing_data['Missing_Count'] > 0].sort_values(
    'Missing_Percentage', ascending=False
)

if len(missing_data) > 0:
    display(missing_data)
    print(f"\nTotal missing values: {df.isnull().sum().sum():,}")
else:
    print("No missing values found!")
```

```python
# Check datetime range
print("7. Temporal Coverage:")
print(f"Start date: {df['datetime'].min()}")
print(f"End date: {df['datetime'].max()}")
print(f"Duration: {(df['datetime'].max() - df['datetime'].min()).days} days")
print(f"Frequency: Approximately {len(df) / ((df['datetime'].max() - df['datetime'].min()).days):.0f} records per day")
```

```python
# Check for duplicates
print("\n8. Duplicate Records:")
duplicates = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicates:,}")

if duplicates > 0:
    print(f"Percentage of duplicates: {(duplicates / len(df)) * 100:.2f}%")
```

```python
# Save the loaded dataframe for use in other notebooks
print("9. Saving initial loaded data...")
df.to_pickle(os.path.join(PROCESSED_PATH, 'df_loaded.pkl'))
print("Data saved to: data/processed/df_loaded.pkl")
```
